**FEATURE ENGINEERING**

```python
#FEATURE ENGINEERING - RFM
#RFM Scores(Recency, Frequency, Monetary)

# 1. Define Snapshot Date for Recency Calculation
snapshot_date = df['order_purchase_timestamp'].max() + pd.Timedelta(days=1)
print(snapshot_date)
```

```python
# RFM
# 1. Create RFM Table using GroupBy Aggregation
# Re-calculate RFM from the original 'df' DataFrame
rfm = df.groupby('customer_unique_id').agg(
    Recency=('order_purchase_timestamp', lambda x: (snapshot_date - x.max()).days),
    Frequency=('order_id', 'nunique'),
    Monetary=('total_amount', 'sum')
).reset_index()


# 2. Rename Columns to RFM Metrics (already done implicitly with new agg syntax, but keeping for clarity)
rfm.columns = ['customer_unique_id', 'Recency', 'Frequency', 'Monetary']


# 3. Display First Few Rows of RFM Table
rfm.head()
```

```python
#RFM Score
# 1. Assign Recency Score (Lower Recency = Higher Score)

rfm['R_score'] = pd.qcut(rfm['Recency'], 4, labels=[4,3,2,1])


# 2. Assign Frequency Score (Higher Frequency = Higher Score)

rfm['F_score'] = pd.qcut(rfm['Frequency'].rank(method='first'), 4, labels=[1,2,3,4])


# 3. Assign Monetary Score (Higher Spending = Higher Score)

rfm['M_score'] = pd.qcut(rfm['Monetary'], 4, labels=[1,2,3,4])


# 4. Calculate Final RFM Score (Combined Score)

# rfm['RFM_Score'] = rfm[['R_score','F_score','M_score']].sum(axis=1)
rfm['RFM_Score'] = rfm['R_score'].astype(str) + rfm['F_score'].astype(str) + rfm['M_score'].astype(str)

print(rfm['RFM_Score'])
```

```python
#FEATURE ENGINEERING — TIME SERIES (ROLLING)

# -------------------------------
# 1. Extract Date from Timestamp (Remove Time Component)
# -------------------------------
df['order_purchase_timestamp'] = pd.to_datetime(df['order_purchase_timestamp'], errors = 'coerce').dt.date

# -------------------------------
# 2. Aggregate Total Sales per Day
# -------------------------------
daily_sales = df.groupby('order_purchase_timestamp', as_index=False).agg({
    'total_amount': 'sum'
})

# -------------------------------
# 3. Convert Date Column Back to Datetime Format
# -------------------------------
daily_sales['order_purchase_timestamp'] = pd.to_datetime(daily_sales['order_purchase_timestamp'])

# -------------------------------
# 4. Preview Daily Sales Data
# -------------------------------
print(daily_sales.head())

print()
print()
# -------------------------------
# 5. Sort Data by Date (Chronological Order)
# -------------------------------
daily_sales = daily_sales.sort_values('order_purchase_timestamp')
print(daily_sales)
```

```python
#ROLLING FEATURES

# 1. Calculate 7-Day Rolling Mean (Short-Term Trend)

daily_sales['rolling_mean_7'] = daily_sales['total_amount'].rolling(7).mean()


# 2. Calculate 30-Day Rolling Mean (Long-Term Trend)

daily_sales['rolling_mean_30'] = daily_sales['total_amount'].rolling(30).mean()
print(daily_sales)
```

**K-MEANS CLUSTERING**

```python
#Select Features


# 1. Select RFM Features for Modeling/Clustering

rfm_data = rfm[['Recency', 'Frequency', 'Monetary']]
print(rfm_data)
```

```python
#Log Transformation(Because Monetary is highly skewed)

# 1. Apply Log Transformation to RFM Data (Reduce Skewness)
import numpy as np

rfm_log = np.log1p(rfm_data)
print(rfm_log)


#Scaling
# -------------------------------
# 1. Import StandardScaler for Feature Scaling
# -------------------------------
from sklearn.preprocessing import StandardScaler


# -------------------------------
# 2. Initialize Scaler Object
# -------------------------------
scaler = StandardScaler()


# -------------------------------
# 3. Apply Scaling to Log-Transformed RFM Data
# -------------------------------
rfm_scaled = scaler.fit_transform(rfm_log)
print(rfm_scaled)
```

```python
#Find Optimal Clusters (Elbow Method)
import os
os.environ["LOKY_MAX_CPU_COUNT"] = "4"
# -------------------------------
# 1. Import Required Libraries (KMeans & Plotting)
# -------------------------------
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt


# -------------------------------
# 2. Initialize Inertia List (Store WCSS Values)
# -------------------------------
inertia = []


# -------------------------------
# 3. Run K-Means for Different Values of K (2 to 9)
# -------------------------------
for k in range(2, 10):
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(rfm_scaled)
    inertia.append(kmeans.inertia_)


# -------------------------------
# 4. Plot Elbow Curve (K vs Inertia)
# -------------------------------
plt.plot(range(2,10), inertia, marker='o')
plt.xlabel('Number of Clusters')
plt.ylabel('Inertia')
plt.title('Elbow Method')
plt.show()
```

```python
# -------------------------------
# 1. Import Required Libraries (KMeans & Evaluation Metric)
# -------------------------------
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score


# -------------------------------
# 2. Test Multiple K Values (2 to 7) for Optimal Clustering
# -------------------------------
for k in range(2, 7):

    # Initialize KMeans with K clusters
    kmeans = KMeans(n_clusters=k, random_state=42)

    # Fit model and predict cluster labels
    labels = kmeans.fit_predict(rfm_scaled)

    # Calculate Silhouette Score for current K
    score = silhouette_score(rfm_scaled, labels)

    # Print results
    print(f"K={k}, Silhouette Score={score}")
```

```python
# -------------------------------
# 1. Initialize K-Means Model with Selected Number of Clusters (K=6)
# -------------------------------
kmeans = KMeans(n_clusters=6, random_state=42)


# -------------------------------
# 2. Fit Model and Assign Cluster Labels to Customers
# -------------------------------
rfm['Cluster_KMeans'] = kmeans.fit_predict(rfm_scaled)
```

```python
# -------------------------------
# 1. Analyze Cluster Characteristics (Mean RFM Values per Cluster)
# -------------------------------
rfm.groupby('Cluster_KMeans')[['Recency','Frequency','Monetary']].mean()
```

**CLUSTER EVALUATION For K-Mean**

```python
# -------------------------------
# 1. Calculate Silhouette Score for K-Means Clustering
# -------------------------------
score_kmeans = silhouette_score(rfm_scaled, rfm['Cluster_KMeans'])


# -------------------------------
# 2. Display K-Means Clustering Score
# -------------------------------
print("K-Means Score:", score_kmeans)
```

```python
#INSTALL & IMPORT PROPHET
# -------------------------------
# 1. Install Prophet Library (Run in Terminal / Notebook Cell)
# -------------------------------
%pip install prophet


# -------------------------------
# 2. Import Prophet Model
# -------------------------------
from prophet import Prophet
```

```python

# -------------------------------
# 1. Ensure Timestamp Column is in Datetime Format
# -------------------------------
df['order_purchase_timestamp'] = pd.to_datetime(df['order_purchase_timestamp'])


# -------------------------------
# 2. Extract Date (Remove Time Component)
# -------------------------------
df['order_date'] = df['order_purchase_timestamp'].dt.date


# -------------------------------
# 3. Aggregate Daily Sales (Total Amount per Day)
# -------------------------------
daily_sales = df.groupby('order_date')['total_amount'].sum().reset_index()


# -------------------------------
# 4. Rename Columns for Prophet (ds = date, y = target)
# -------------------------------
prophet_df = daily_sales.rename(columns={
    'order_date': 'ds',
    'total_amount': 'y'
})

prophet_df['ds'] = pd.to_datetime(prophet_df['ds'])

# ---------------------------------------------------------
# Define Active Dataset Based on Cutoff Date
# ---------------------------------------------------------
active_df = prophet_df[prophet_df['ds'] <= '2018-08-31']


# ---------------------------------------------------------
# Compute Threshold for Bottom 5% Values
# ---------------------------------------------------------
threshold = active_df['y'].quantile(0.05)


# ---------------------------------------------------------
# Remove Low-Volume Noise Below Threshold
# ---------------------------------------------------------
active_df = active_df[active_df['y'] > threshold]
```

**MODEL EVALUATION**

**Train-Test Split**

```python
# Split Data into Training and Testing Sets

# -------------------------------
# Handle Outliers using 99th Percentile Clipping
# -------------------------------
cap = active_df['y'].quantile(0.99)


# ---------------------------------------------------------
# Clip Extreme Outliers Above Cap Value
# ---------------------------------------------------------
active_df['y'] = active_df['y'].clip(upper=cap)

# ---------------------------------------------------------
# Apply Log Transformation to Stabilize Target Variable
# ---------------------------------------------------------
active_df['y'] = np.log1p(active_df['y'])


# -------------------------------
# Split Data into Training Set (All Except Last 30 Days)
# -------------------------------
train = active_df.iloc[:-30]
print("Train :", len(train))


# -------------------------------
# Split Data into Testing Set (Last 30 Days)
# -------------------------------
test = active_df.iloc[-30:]
print("Test :", len(test))
```

```python
# ----------------------------------------------------------------------------
# Initialize Prophet Model with Multiplicative Seasonality & Tuning Parameters
# ----------------------------------------------------------------------------
prophet_model = Prophet(
    holidays_prior_scale=20,
    changepoint_prior_scale=0.1, # Increased for better flexibility to non-stationary data
    seasonality_prior_scale=15,
    seasonality_mode='multiplicative', # Multiplicative is best for growing retail sales
    yearly_seasonality=True,
    weekly_seasonality=True,
    daily_seasonality=False # Set to False for daily-aggregated data
)

# -------------------------------
# 2. Add Country-Specific Holidays (BRAZIL)
# -------------------------------
prophet_model.add_country_holidays(country_name='BR')

# -------------------------------
# 3. Add Monthly Seasonality to Prophet Model
# -------------------------------
prophet_model.add_seasonality(name='monthly', period=30.5, fourier_order=10)


# -------------------------------
# 4. Train Model on Training Data
# -------------------------------
prophet_model.fit(train)


# -------------------------------
# 5. Create Future DataFrame (Next 30 Days)
# -------------------------------
future = prophet_model.make_future_dataframe(periods=30)


# -------------------------------
# 6. Generate Forecast Predictions
# -------------------------------
forecast = prophet_model.predict(future)


# -------------------------------
# 7. Display Last 30 Forecasted Values
# -------------------------------
forecast[['ds','yhat']].tail()
```

```python
#Visualization

# -------------------------------
# 1. Plot Overall Sales Forecast
# -------------------------------
fig = prophet_model.plot(forecast)
plt.xlabel("Date")
plt.ylabel("Sales")
plt.title("Sales Forecast using Prophet")
plt.show()
```

```python
# -------------------------------
# 2. Plot Forecast Components (Trend + Seasonality)
# -------------------------------
prophet_model.plot_components(forecast)
```

```python
# -------------------------------
# 1. Align Forecast with Test Dates
# -------------------------------
pred = forecast[forecast['ds'].isin(test['ds'])]

merged = test[['ds','y']].merge(pred[['ds','yhat']], on='ds', how='inner')



# -------------------------------
# 5. Preview Final Results
# -------------------------------
print(merged[['y','yhat']].head(20))
```

```python
# -------------------------------
# 1. Import Evaluation Metric (MAE)
# -------------------------------
from sklearn.metrics import mean_absolute_error

# -------------------------------
# 2. Calculate Mean Absolute Error (MAE)
# -------------------------------
mae = mean_absolute_error(merged['y'], merged['yhat'])


# -------------------------------
# 3. Display Model Performance
# -------------------------------
print("MAE:", mae)
```

```python
# -------------------------------
# 1. Import Required Libraries (RMSE Calculation)
# -------------------------------
from sklearn.metrics import mean_squared_error

# -------------------------------
# 2. Calculate Root Mean Squared Error (RMSE)
# -------------------------------
rmse = np.sqrt(mean_squared_error(merged['y'], merged['yhat']))


# -------------------------------
# 3. Display RMSE Value
# -------------------------------
print("RMSE:", rmse)
```

```python

#Mean Absolute Percentage Error (MAPE)

# -------------------------------
# 1. Calculate Mean Absolute Percentage Error (MAPE)
# -------------------------------
mape = np.mean(np.abs((merged['y'] - merged['yhat']) / merged['y'])) * 100

# -------------------------------
# 2. Display MAPE Value
# -------------------------------
print("MAPE:", mape)
```

```python
#PREPARE DATA FOR LSTM

import numpy as np
daily_sales['order_date'] = pd.to_datetime(daily_sales['order_date'])

daily_resampled = daily_sales.set_index('order_date').resample('D').sum().fillna(0)

# 1. Filter out the anomalous 'dead zone' (data after 2018-08-31)
daily_sales_filtered = daily_resampled[daily_resampled.index <= '2018-08-31']


# 2. Apply log transformation to stabilize variance and improve LSTM convergence
data = np.log1p(daily_sales_filtered[['total_amount']].values)

print(data[:5]) # Displaying first 5 log-scaled values
```

```python

train_size = len(data) - 30
# -------------------------------
# Split Data FIRST to prevent data leakage
# -------------------------------
train_data = data[:train_size]
test_data = data[train_size:]


# -------------------------------
# Import MinMaxScaler
# -------------------------------
from sklearn.preprocessing import MinMaxScaler, StandardScaler


# -------------------------------
# Initialize Scaler
# -------------------------------
# scaler = MinMaxScaler(feature_range=(0, 1))
scaler = StandardScaler()

# -------------------------------
# Transform Train and Test Data
# -------------------------------
train_scaled = scaler.fit_transform(train_data)
test_scaled = scaler.transform(test_data)


# -------------------------------
# Check Scaled Output
# -------------------------------
print(train_scaled[:5])
```

```python
#Create Sequences
import torch
# -------------------------------
# 1. Define Sequence Creation Function
# -------------------------------
def create_sequences(data, seq_length=30):
    X, y = [], []

    for i in range(len(data) - seq_length):
        X.append(data[i:i+seq_length])
        y.append(data[i+seq_length])

    return torch.tensor(np.array(X), dtype=torch.float32), torch.tensor(np.array(y), dtype=torch.float32)


# -------------------------------
# 2. Create Sequences for Training Data
# -------------------------------
X_train, y_train = create_sequences(train_scaled)


# -------------------------------
# 3. Extract Last 30 Days from Training Data (Context for Test)
# -------------------------------
last_30_days = train_scaled[-30:]


# -------------------------------
# 4. Combine Training Context with Test Data
# -------------------------------
test_combined = np.vstack((last_30_days, test_scaled))


# -------------------------------
# 5. Create Sequences for Testing Data
# -------------------------------
X_test, y_test = create_sequences(test_combined)


# -------------------------------
# 6. Verify Shape of Test Sequences
# -------------------------------
print(X_test.shape)
```

```python
# -------------------------------
# Import PyTorch Lightning Library (High-Level Training Framework)
# -------------------------------
import pytorch_lightning as pl


# -------------------------------
# Import Neural Network Modules (Layers, Loss Functions)
# -------------------------------
import torch.nn as nn


# -------------------------------
# Define LSTM Model using PyTorch Lightning
# -------------------------------
class LSTMModel(pl.LightningModule):

    def __init__(self):
        super().__init__()

        # LSTM Layer
        self.lstm = nn.LSTM(1, 128, num_layers = 1, batch_first=True)


        # Fully Connected Layer
        self.fc = nn.Linear(128, 1)

        # Loss Function
        self.loss_fn = nn.MSELoss()

    # -------------------------------
    # Forward Pass
    # -------------------------------
    def forward(self, x):
        out, _ = self.lstm(x)
        out = self.fc(out[:, -1, :])  # Last time step
        return out

    # -------------------------------
    # Training Step
    # -------------------------------
    def training_step(self, batch, batch_idx):
        x, y = batch
        y_hat = self(x)
        loss = self.loss_fn(y_hat, y)

        # self.log("train_loss", loss, prog_bar=True)
        return loss

    # -------------------------------
    # Optimizer
    # -------------------------------
    def configure_optimizers(self):
        return torch.optim.Adam(self.parameters(), lr=0.005)
```

```python
# -------------------------------
# Import Dataset and DataLoader Utilities
# -------------------------------
from torch.utils.data import TensorDataset, DataLoader

torch.manual_seed(42)
# -------------------------------
# Initialize LSTM Model
# -------------------------------
lstm_model = LSTMModel()


# -------------------------------
# Create Training Dataset (Features + Target)
# -------------------------------
train_dataset = TensorDataset(X_train, y_train)


# -------------------------------
# Create DataLoader for Batch Processing
# -------------------------------
train_loader = DataLoader(train_dataset, batch_size=16, shuffle=True)
```

```python

# -------------------------------
# 2. Initialize PyTorch Lightning Trainer
# -------------------------------
trainer = pl.Trainer(
    max_epochs=150,
    gradient_clip_val=0.5,
    accelerator="auto",   # Uses GPU if available
    devices=1,
    enable_checkpointing=False,
    logger=False
)


# -------------------------------
# 3. Train the Model
# -------------------------------
trainer.fit(lstm_model, train_loader)
```

```python
#Define objective function

# -------------------------------
# Import Libraries
# -------------------------------
import optuna
from sklearn.metrics import roc_auc_score
from xgboost import XGBClassifier
from sklearn.model_selection import train_test_split


# -------------------------------
# Create Train-Validation Split
# -------------------------------
X_tr, X_val, y_tr, y_val = train_test_split(
    X_train, y_train, test_size=0.2, random_state=42
)


# -------------------------------
# Define Objective Function
# -------------------------------
def objective(trial):

    params = {
        "n_estimators": trial.suggest_int("n_estimators", 50, 300),
        "max_depth": trial.suggest_int("max_depth", 3, 10),
        "learning_rate": trial.suggest_float("learning_rate", 0.01, 0.3),
        "subsample": trial.suggest_float("subsample", 0.5, 1.0),
        "colsample_bytree": trial.suggest_float("colsample_bytree", 0.5, 1.0)
    }

    model = XGBClassifier(**params, random_state=42)

    # -------------------------------
    # Use TRAIN subset
    # -------------------------------
    model.fit(X_tr, y_tr)

    # -------------------------------
    # Evaluate on VALIDATION set
    # -------------------------------
    y_prob = model.predict_proba(X_val)[:, 1]
    auc = roc_auc_score(y_val, y_prob)

    return auc
```

```python
#Run optimization

# -------------------------------
# Create Optuna Study for Hyperparameter Optimization
# -------------------------------
study = optuna.create_study(direction="maximize")


# -------------------------------
# Run Optimization Trials to Maximize AUC Score
# -------------------------------
study.optimize(objective, n_trials=20)
```

```python
#Best parameters

# -------------------------------
# Display Best Hyperparameters from Optuna Optimization
# -------------------------------
print("Best Parameters:", study.best_params)


# -------------------------------
# Display Best Achieved AUC Score
# -------------------------------
print("Best AUC:", study.best_value)
```

**TRAIN FINAL MODEL**

```python
# -------------------------------
# Initialize Final Model with Optimized Hyperparameters
# -------------------------------
best_model = XGBClassifier(**study.best_params, random_state=42)


# -------------------------------
# Train Final Optimized Model on Full Training Data
# -------------------------------
best_model.fit(X_train, y_train)
```

**AIRFLOW DAG**

```python
from airflow import DAG
from airflow.operators.python import PythonOperator
from datetime import datetime
import pandas as pd
import os
from xgboost import XGBClassifier
import joblib

# Ensure directories exist
os.makedirs("models", exist_ok=True)
os.makedirs("temp", exist_ok=True)

# -------------------------------
# 1. Data Loading Function
# -------------------------------
def load_data():
    df = pd.read_csv("dataset/cleaned_dataset.csv") # Updated path to match your root
    df.to_csv("dataset/temp/data.csv", index=False)

# -------------------------------
# 2. Data Preprocessing Function
# -------------------------------
def preprocess_data():
    df = pd.read_csv("dataset/temp/data.csv")
    # Feature Engineering (Ensuring columns exist)
    if 'price' in df.columns and 'freight_value' in df.columns:
        df['total_amount'] = df['price'] + df['freight_value']
    df.to_csv("dataset/temp/processed.csv", index=False)

# -------------------------------
# 3. Model Training Function
# -------------------------------
def train_model():
    df = pd.read_csv("dataset/temp/processed.csv")

    # Selecting available numeric features
    cols = [c for c in ['price', 'freight_value', 'total_amount'] if c in df.columns]
    X = df[cols]
    y = (df[cols[0]] > df[cols[0]].median()).astype(int) # Dummy target for logic

    model = XGBClassifier()
    model.fit(X, y)

    joblib.dump(model, "models/churn_model.pkl")

# -------------------------------
# 4. DAG Configuration (FIXED)
# -------------------------------
default_args = {
    'owner': 'airflow',
    'start_date': datetime(2024, 1, 1)
}

dag = DAG(
    'retraining_pipeline',
    default_args=default_args,
    schedule='@daily',  # <--- CHANGED FROM schedule_interval to schedule
    catchup=False
)

# -------------------------------
# 5. Create Airflow Tasks
# -------------------------------
load_task = PythonOperator(
    task_id='load_data',
    python_callable=load_data,
    dag=dag
)

preprocess_task = PythonOperator(
    task_id='preprocess_data',
    python_callable=preprocess_data,
    dag=dag
)

train_task = PythonOperator(
    task_id='train_model',
    python_callable=train_model,
    dag=dag
)

# -------------------------------
# 6. Pipeline Flow
# -------------------------------
load_task >> preprocess_task >> train_task
```
